## Statistical Analysis

```python
# Load Dataset

df = pd.read_csv("../01_dataset/Superstore.csv", encoding="latin1")

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)
```

```python
# Descriptive Statistics

print("===== DESCRIPTIVE STATISTICS =====")

print("\nMean:")
print(df[['Sales', 'Quantity', 'Discount', 'Profit']].mean().round(2))

print("\nMedian:")
print(df[['Sales', 'Quantity', 'Discount', 'Profit']].median().round(2))

print("\nStandard Deviation:")
print(df[['Sales', 'Quantity', 'Discount', 'Profit']].std().round(2))
```

```python
# Hypothesis Testing: Discount vs Profit

from scipy.stats import ttest_ind

# Two groups
profit_no_discount = df[df['Discount'] == 0]['Profit']
profit_with_discount = df[df['Discount'] > 0]['Profit']

# Independent t-test
t_stat, p_value = ttest_ind(
    profit_no_discount,
    profit_with_discount,
    equal_var=False
)

print("===== HYPOTHESIS TESTING =====")
print("T-Statistic:", round(t_stat, 2))
print("P-Value:", p_value)

if p_value < 0.05:
    print("\nResult: Significant difference in profit.")
else:
    print("\nResult: No significant difference in profit.")
```

```python
# ANOVA Test: Category vs Profit

from scipy.stats import f_oneway

furniture_profit = df[df['Category'] == 'Furniture']['Profit']
office_profit = df[df['Category'] == 'Office Supplies']['Profit']
technology_profit = df[df['Category'] == 'Technology']['Profit']

f_stat, p_value = f_oneway(
    furniture_profit,
    office_profit,
    technology_profit
)

print("===== ANOVA TEST =====")
print("F-Statistic:", round(f_stat, 2))
print("P-Value:", p_value)

if p_value < 0.05:
    print("\nResult: Significant difference in average profit between categories.")
else:
    print("\nResult: No significant difference in average profit between categories.")
```

## Statistical Analysis Conclusion

The statistical analysis helped us understand the distribution and relationships
within the Superstore dataset.

### Key Findings

1. Mean, Median and Standard Deviation were used to understand the central
   tendency and variation of Sales, Quantity, Discount and Profit.

2. Independent T-Test was used to compare Profit between orders with and
   without Discount.

3. ANOVA Test was used to compare average Profit across Furniture,
   Office Supplies and Technology categories.

4. Hypothesis testing helps determine whether the observed differences
   are statistically significant rather than occurring by random chance.

### Business Interpretation

Statistical analysis provides evidence that can support business decisions.
The company should evaluate discount strategies and category-level
profitability rather than focusing only on total sales.

Overall, statistical analysis confirms that data-driven decisions can help
identify important differences in business performance.

##  Machine Learning

### Objective

The objective of this section is to build a Machine Learning model
that can predict Profit based on available business features.

We will:
- Prepare the data for Machine Learning
- Select features and target
- Split data into training and testing sets
- Train a Machine Learning model
- Evaluate model performance
- Interpret the results

```python
# ML: Feature and Target Selection

X = df[['Sales', 'Quantity', 'Discount']]
y = df['Profit']

print("Features (X):")
print(X.head())

print("\nTarget (y):")
print(y.head())

print("\nFeature Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
# Train-Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training Data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting Data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
# Linear Regression Model

from sklearn.linear_model import LinearRegression

model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print("Linear Regression Model trained successfully!")
```

```python
# Make Predictions

y_pred = model.predict(X_test)

print("===== SAMPLE PREDICTIONS =====")
print("Actual Profit:")
print(y_test.head(10).values)

print("\nPredicted Profit:")
print(y_pred[:10].round(2))
```

```python
# Model Evaluation

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("===== MODEL EVALUATION =====")
print("Mean Absolute Error (MAE):", round(mae, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("R² Score:", round(r2, 4))
```

```python
# Actual vs Predicted Profit

plt.figure(figsize=(8, 5))

plt.scatter(y_test, y_pred, alpha=0.5)

plt.xlabel("Actual Profit")
plt.ylabel("Predicted Profit")
plt.title("Actual vs Predicted Profit")

plt.show()
```

```python
# Random Forest Regression

from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("Random Forest Model trained successfully!")
```

```python
# Random Forest Model Evaluation

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("===== RANDOM FOREST EVALUATION =====")
print("MAE:", round(rf_mae, 2))
print("RMSE:", round(rf_rmse, 2))
print("R² Score:", round(rf_r2, 4))
```

```python
# Model Comparison

comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest'],
    'MAE': [mae, rf_mae],
    'RMSE': [rmse, rf_rmse],
    'R2 Score': [r2, rf_r2]
})

print("===== MODEL COMPARISON =====")
print(comparison.round(4))
```

```python
# Feature Importance Analysis

feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

print("===== FEATURE IMPORTANCE =====")
print(feature_importance.round(4))
```

```python
# Feature Importance Visualization

plt.figure(figsize=(8, 5))

plt.bar(
    feature_importance['Feature'],
    feature_importance['Importance']
)

plt.xlabel("Features")
plt.ylabel("Importance")
plt.title("Feature Importance - Random Forest")

plt.show()
```

```python
# Final Machine Learning Conclusion

print("========== FINAL ML CONCLUSION ==========")

print("\n1. Best Model: Random Forest")
print(f"   MAE  : {rf_mae:.2f}")
print(f"   RMSE : {rf_rmse:.2f}")
print(f"   R²   : {rf_r2:.4f}")

print("\n2. Feature Importance:")
for _, row in feature_importance.iterrows():
    print(f"   {row['Feature']}: {row['Importance']:.2%}")

print("\n3. Business Insight:")
print("   Sales is the most important predictor of Profit.")
print("   Discount has the second-highest importance.")
print("   Quantity has comparatively low importance.")

print("\n========== ANALYSIS COMPLETED ==========")
```

## Machine Learning Conclusion

Two regression models were evaluated: Linear Regression and Random Forest.

Random Forest performed better than Linear Regression based on MAE and RMSE.
However, its R² score was negative, indicating that the current three features
(Sales, Quantity and Discount) are not sufficient for strong Profit prediction.

### Key ML Findings

- Random Forest performed better than Linear Regression among the tested models.
- Sales was the most important feature with approximately 79.71% importance.
- Discount contributed approximately 17.28%.
- Quantity contributed approximately 3.01%.
- The negative R² indicates that the current model has limited predictive power.

### Business Interpretation

Sales has the strongest relationship with Profit prediction in the current model.
However, additional business features such as Category, Sub-Category, Region,
Customer Segment and shipping information may be required to build a stronger
Profit prediction model.

Therefore, the ML analysis provides useful insights but the current model
should not be used as a highly accurate Profit forecasting system.
